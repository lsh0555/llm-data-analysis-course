```python
# 컬럼 추가
print(order_items.info())
```

```python
# 컬럼추가(파생 컬럼)
# 라인 합계 금액 : line_total = quantity * unit_price
# print(order_items.info())
order_items["line_total"] = order_items["quantity"] * order_items["unit_price"]

print(order_items.head(10))

print(order_items.info())
```

```python
# orders의 order_id에 중복 없다는 것 확인
print(orders["order_id"].duplicated().sum())
```

```python
# print(order_items.head(10))
# order_status를 함께 출력 하려면 => merge
order_sales = order_items.merge(
    orders[["order_id", "customer_id", "order_date", "order_status"]],
    on = "order_id", how = "left", validate="many_to_one"
)

print(order_sales.head())
```

```python
print(orders.info())

print("order_date 타입 : ", orders["order_date"].dtype)
```

```python
order_sales_test = order_sales.copy()

print(order_sales_test.info())
```

```python
order_sales_test["order_date"] = pd.to_datetime(
    order_sales_test["order_date"]
)

print(order_sales_test.info())
```

```python
order_sales_test2 = order_sales.copy()
print(order_sales_test2.head())
```

```python
# 컬럼으로 찾을 땐 .loc[] 사용
order_sales_test2.loc[0, "order_date"] = "2026-02-30"

print(order_sales_test2.head())    # 0번의 위치인 상위에 존재하기에 .head()로 내용이 잘 들어갔는지 확인한다.
```

```python
order_sales_test2["order_date"] = pd.to_datetime(
    order_sales_test2["order_date"],
    errors="coerce"
)

print(order_sales_test2.info())
print(order_sales_test2.head(5))
print("order_date의 결측값 갯수 : ", order_sales_test2["order_date"].isna().sum())
```
