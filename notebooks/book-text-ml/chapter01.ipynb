### 실습 3. 분석에 필요한 컬럼만 선택하기
 
원본 데이터의 모든 컬럼이 이후 분석에 필요한 것은 아닙니다.
 
이번 과정에서는 다음 컬럼을 먼저 선택합니다.

```python
# 필요한 컬럼만 골라 새 DataFrame 만들기
df_books = df[
    ["순위", "판매상품 ID", "상품명", "판매가",
     "인물", "출판사", "발행(출시)일자", "분야"]
]

# 선택한 컬럼과 앞의 5행 확인하기
print(df_books.columns)
display(df_books.head())
```

### 실행 결과

원본 데이터에서 분석에 필요한 **8개 컬럼**을 선택해 `df_books`를 만들었습니다.

컬럼 이름과 앞의 5행을 확인한 결과, 순위·상품명·판매가·출판사 등 원하는 정보가 표시됩니다.

### 실습 4. 컬럼 이름을 사용하기 쉽게 정리하기

원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.

이번 실습에서는 다음처럼 간단하게 정리할 수 있습니다.

```python
# df_books의 컬럼 이름 변경
df_books = df_books.rename(columns={
    "판매상품 ID": "판매상품ID",
    "인물": "저자",
    "발행(출시)일자": "발행일"
})

# 변경된 컬럼 이름 확인
print(df_books.columns)
```

### 실행 결과

`df_books`의 컬럼 이름 3개를 변경했습니다.

- `판매상품 ID` → `판매상품ID` (공백 제거)
- `인물` → `저자` (의미 명확화)
- `발행(출시)일자` → `발행일` (특수문자 및 긴 이름 간소화)

컬럼 목록을 출력해 새 이름이 적용된 것을 확인했습니다.

### 실습 5. 결측치 확인하기

결측치는 값이 비어 있는 데이터를 의미합니다.

데이터 분석에서는 결측치를 무조건 삭제하지 않습니다.

먼저 어디에 얼마나 있는지 확인하고, 분석 목적에 맞게 처리 방법을 결정합니다.

```python
# 각 컬럼의 결측치 개수
missing = df_books.isna().sum()
print(missing)

# 결측치가 1개 이상인 컬럼만 보기
print(missing[missing > 0])
```

### 실행 결과

`df_books`의 8개 컬럼에서 결측치가 모두 **0개**로 확인되었습니다.

결측치가 있는 컬럼만 따로 조회한 결과도 비어 있습니다.

### 실습 6. 판매가를 숫자로 바꾸기

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.

```python
# 1. 판매가에서 쉼표 제거
df_books["판매가"] = df_books["판매가"].str.replace(",", "", regex=False)

# 2. 숫자형으로 변환
df_books["판매가"] = pd.to_numeric(df_books["판매가"])

# 3. 변환된 자료형 확인
print(df_books["판매가"].dtype)
```

### 실행 결과

`판매가`에서 쉼표를 제거하고 숫자형으로 변환했습니다.

자료형이 `int64`로 표시되어, 이제 판매가를 이용한 계산이 가능합니다.

### 실습 7. 발행일을 날짜형으로 바꾸기

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

```python
# YYYYMMDD 형식의 값을 날짜 자료형으로 변환
df_books["발행일"] = pd.to_datetime(
    df_books["발행일"], format="%Y%m%d"
)

# 자료형과 앞의 5개 값 확인
print(df_books["발행일"].dtype)
print(df_books["발행일"].head())
```

### 실행 결과

`발행일`을 날짜 자료형(`datetime64[us]`)으로 변환했습니다.

앞의 5개 값도 `2014-05-19`처럼 연도·월·일 형태로 표시됩니다.

### 실습 8. 문자열 앞뒤 공백 정리하기

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.

```python
# 공백을 제거할 컬럼 이름
text_columns = ["상품명", "저자", "출판사", "분야"]

# 각 컬럼 값의 앞뒤 공백 제거
for column in text_columns:
    df_books[column] = df_books[column].str.strip()

# 처리한 컬럼의 앞 5행 확인
display(df_books[text_columns].head())
```

### 실행 결과

`상품명`, `저자`, `출판사`, `분야` 컬럼의 문자열에서 앞뒤 공백을 제거했습니다.

처리 후 네 컬럼의 앞 5행을 확인했습니다.

### 실습 9. 중복 데이터 확인하기

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.
 
하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

우선 상품을 구분하는 판매상품ID를 확인합니다.

```python
# 앞에 나온 행과 중복되는 행의 개수만 확인합니다.
print("전체 행 중복:", df_books.duplicated().sum())
print("판매상품ID 중복:", df_books.duplicated(subset=["판매상품ID"]).sum())
print("상품명 중복:", df_books.duplicated(subset=["상품명"]).sum())
```

### 실행 결과

전체 행, `판매상품ID`, `상품명`을 기준으로 확인한 중복 행은 각각 **0개**였습니다.

중복 행을 삭제하거나 데이터를 변경하지는 않았습니다.
